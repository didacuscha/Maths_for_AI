```python
def symmetric_matrix(matriza):
    vals, vecs = np.linalg.eig(matriza)
    # we have to transform eigenvalues to diagonal 
    vals_diag = np.diag(vals)
    # we have to get the inverse of the eigenvectors
    vecs_inverse = np.linalg.inv(vecs)
    A = np.dot(np.dot(vecs, vals_diag), vecs_inverse)
    return A
    
```

```python
matrix_a = np.array([[2, 1],[1, 2]])
matrix_b = np.array([[2, 0],[0, 2]])
matrix_c = np.array([[2, 1],[0, 2]])

a = symmetric_matrix(matrix_a)
b = symmetric_matrix(matrix_b)
c = symmetric_matrix(matrix_c)

print(a,'\n \n ', b, '\n \n ', c)
```
