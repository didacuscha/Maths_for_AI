```python
def plot_decision_boundary(w, scatter_plot, data):
    """
    Plots the decision boundary line w₁x₁ + w₂x₂ + w₃ = 0 on the scatter plot.
    
    Parameters:
    - w: list or numpy array with 3 elements [w₁, w₂, w₃]
    - scatter_plot: the existing scatter plot
    - data: pandas DataFrame containing the data
    """
    # Extract the axis limits from the current plot
    x_min, x_max = plt.xlim()
    
    # Create points for the line
    x1_values = np.linspace(x_min, x_max, 100)
    
    # Calculate corresponding x2 values using the equation w₁x₁ + w₂x₂ + w₃ = 0
    # Solving for x2: x2 = (-w₁x₁ - w₃) / w₂
    x2_values = (-w[0] * x1_values - w[2]) / w[1]
    
    # Plot the decision boundary
    plt.plot(x1_values, x2_values, 'r-', linewidth=2, label=f'Decision Boundary: {w[0]:.4f}x₁ + {w[1]:.4f}x₂ + {w[2]:.4f} = 0')
    
    # Update the legend
    plt.legend(title='Prognosis (0: Benign, 1: Malignant)')
    
    plt.tight_layout()
    return plt
```

```python
def calculate_fw(w, X, t):
    """
    Calculate the value of f(w) to quantify how good w is for classifying the training data.
    
    Parameters:
    - w: list or numpy array with 3 elements [w₁, w₂, w₃]
    - X: feature matrix where each row is a data point [x₁, x₂, 1]
    - t: labels (0 for benign, 1 for malignant)
    
    Returns:
    - f(w): a measure of classification quality (lower is better)
    """
    # Ensure w is a numpy array
    w = np.array(w)
    
    # Prepare the data matrix by adding a column of 1s for the bias term
    # If X doesn't already have the bias term
    if X.shape[1] == 2:
        X_with_bias = np.column_stack((X, np.ones(X.shape[0])))
    else:
        X_with_bias = X
        
    # Calculate z = X·w
    z = np.dot(X_with_bias, w)
    
    # Calculate sigmoid(z)
    sigmoid_z = 1 / (1 + np.exp(-z))
    
    # Calculate cross-entropy loss
    epsilon = 1e-15  # Small value to avoid log(0)
    loss = -np.mean(t * np.log(sigmoid_z + epsilon) + (1 - t) * np.log(1 - sigmoid_z + epsilon))
    
    return loss
```

```python
def evaluate_candidates(data, candidate_weights):
    """
    Evaluate and compare the candidate weight vectors.
    
    Parameters:
    - data: pandas DataFrame containing the breast cancer data
    - candidate_weights: list of weight vectors to evaluate
    """
    # Extract features and target
    X = data[['radius_mean', 'smoothness_mean']].values
    t = data['pronostic'].values
    
    # Create scatter plot
    plt.figure(figsize=(12, 10))
    scatter_plot = sns.scatterplot(
        data=data,
        x='radius_mean',
        y='smoothness_mean',
        hue='pronostic', 
        palette='viridis',
        style='pronostic',
        s=70
    )
    
    # Add titles and labels
    plt.title('Breast Cancer Data: Radius Mean vs. Smoothness Mean with Decision Boundaries')
    plt.xlabel('Radius Mean')
    plt.ylabel('Smoothness Mean')
    plt.grid(True)
    
    # Results storage
    results = []
    
    # Evaluate each candidate
    colors = ['r', 'g', 'b']  # Different colors for different boundaries
    
    for i, w in enumerate(candidate_weights):
        # Calculate f(w)
        fw_value = calculate_fw(w, X, t)
        results.append((w, fw_value))
        
        # Plot decision boundary
        x_min, x_max = plt.xlim()
        x1_values = np.linspace(x_min, x_max, 100)
        x2_values = (-w[0] * x1_values - w[2]) / w[1]
        
        plt.plot(x1_values, x2_values, color=colors[i], linestyle='-', linewidth=2, 
                 label=f'w{i+1}: f(w)={fw_value:.4f}')
    
    # Update legend and show plot
    plt.legend(title='Legend')
    plt.tight_layout()
    plt.show()
    
    # Print results
    print("Evaluation of candidate weight vectors:")
    for i, (w, fw) in enumerate(results):
        print(f"Candidate {i+1}: w = {w}, f(w) = {fw:.6f}")
    
    # Identify the best candidate
    best_idx = min(range(len(results)), key=lambda i: results[i][1])
    print(f"\nBest candidate is #{best_idx+1} with f(w) = {results[best_idx][1]:.6f}")
    
    return results
```

```python
# candidatos w planteados en el taller
candidate_weights = [
    [0.0603, 4.0364, 1.3192],
    [-1.345, 3.597, -0.588],
    [0.0603, 1.000, 0.502]]


resultados = evaluate_candidates(data, candidate_weights)
```

```python
def calculate_gradient_and_hessian(w, X, t):
    """
    Calculate the gradient and Hessian matrix for logistic regression.
    
    Parameters:
    - w: weight vector [w₁, w₂, w₃]
    - X: feature matrix where each row is a data point [x₁, x₂]
    - t: target labels (0: benign, 1: malignant)
    
    Returns:
    - gradient: gradient vector of f(w)
    - hessian: Hessian matrix of f(w)
    """    
    # Ensure w is a numpy array
    w = np.array(w)
    
    # Prepare the data matrix by adding a column of 1s for the bias term
    if X.shape[1] == 2:
        X_with_bias = np.column_stack((X, np.ones(X.shape[0])))
    else:
        X_with_bias = X
    
    n_samples = X_with_bias.shape[0]
    n_features = X_with_bias.shape[1]
    
    # Calculate predictions using sigmoid function
    z = np.dot(X_with_bias, w)
    y_pred = 1 / (1 + np.exp(-z))
    
    # Calculate error
    error = y_pred - t
    
    # Calculate gradient: gradient = X^T * (y_pred - t) / n_samples
    gradient = np.dot(X_with_bias.T, error) / n_samples
    
    # Calculate Hessian matrix: H = X^T * diag(y_pred * (1 - y_pred)) * X / n_samples
    S = np.diag(y_pred * (1 - y_pred))
    hessian = np.dot(X_with_bias.T, np.dot(S, X_with_bias)) / n_samples
    
    return gradient, hessian

def analyze_weight_vectors(data, candidate_weights):
    """
    Analyze each candidate weight vector by calculating its gradient and Hessian.
    
    Parameters:
    - data: pandas DataFrame containing the breast cancer data
    - candidate_weights: list of weight vectors to evaluate
    """
    # Extract features and target
    X = data[['radius_mean', 'smoothness_mean']].values
    t = data['pronostic'].values
    
    print("Analysis of candidate weight vectors:")
    
    for i, w in enumerate(candidate_weights):
        # Calculate gradient and Hessian
        gradient, hessian = calculate_gradient_and_hessian(w, X, t)
        
        # Calculate eigenvalues of Hessian
        eigenvalues = np.linalg.eigvals(hessian)
        
        # Calculate gradient magnitude
        gradient_magnitude = np.linalg.norm(gradient)
        
        print(f"\nCandidate {i+1}: w = {w}")
        print(f"Gradient: {gradient}")
        print(f"Gradient magnitude: {gradient_magnitude:.6f}")
        print(f"Hessian matrix:\n{hessian}")
        print(f"Hessian eigenvalues: {eigenvalues}")
        print(f"Is Hessian positive definite: {all(eigenvalues > 0)}")
```

```python
analyze_weight_vectors(data, candidate_weights)
```
